# Run `record_batch.py` with tmux

This notebook can execute tmux commands directly.

```python
import subprocess
import shlex


session = "Train"
script_dir = "~/Workspace/safe_bigym_hoi/external/robobase/"
script = "train.py"
log_file = f"{script_dir}/train.log"
args = [
    "method=act",
    "env=bigym/drawer_top_open.yaml",
    "launch=act_pixel_bigym",
]

def run(cmd: str):
    print(f"$ {cmd}")
    return subprocess.run(cmd, shell=True, check=False, text=True, capture_output=False)
```

## Start recording in detached tmux session

```python
inner_cmd = (
    f"cd {shlex.quote(script_dir)} && "
    f"python3 {shlex.quote(script)} "
    + " ".join(shlex.quote(arg) for arg in args)
    + f" 2>&1 | tee {shlex.quote(log_file)}"
)

cmd = f"tmux new -d -s {shlex.quote(session)} {shlex.quote(inner_cmd)}"
run(cmd)
```

## List tmux sessions

```python
run("tmux ls")
```

## Attach to session (run in terminal if notebook UI cannot attach interactive TTY)

```python
# Run the print output in terminal
print(f"tmux attach -t {session}")
```

Detach without stopping: `Ctrl-b` then `d`.

## Tail the log

```python
num_of_line_display = 4
run(f"tail -n {num_of_line_display} {log_file}")
```

## Kill session

```python
run(f"tmux kill-session -t {session}")
```
